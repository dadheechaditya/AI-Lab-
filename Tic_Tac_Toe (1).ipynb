```python

# Tic-Tac-Toe

board = [' '] * 9

def display():
    print()
    for i in range(0, 9, 3):
        print(" | ".join(board[i:i+3]))
        if i < 6:
            print("--+---+--")
    print()

def check_winner(p):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]

    for a, b, c in wins:
        if board[a] == board[b] == board[c] == p:
            return True
    return False

player = 'X'

for turn in range(9):
    display()
    print("Player", player)

    move = int(input("Enter position (1-9): ")) - 1

    if move < 0 or move > 8 or board[move] != ' ':
        print("Invalid move! Try again.")
        continue

    board[move] = player

    if check_winner(player):
        display()
        print(player, "wins!")
        break

    player = 'O' if player == 'X' else 'X'

else:
    display()
    print("It's a draw!")
```
